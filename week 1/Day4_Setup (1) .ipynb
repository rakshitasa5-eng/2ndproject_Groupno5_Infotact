```python
import pandas as pd
```

```python
customer = pd.read_csv("DATA/RAW/Customer.csv")
employee = pd.read_csv("DATA/RAW/Employee.csv")
orders = pd.read_csv("DATA/RAW/Orders.csv")
orderdetails = pd.read_csv("DATA/RAW/OrderDetails.csv")
product = pd.read_csv("DATA/RAW/Product.csv")
warehouse = pd.read_csv("DATA/RAW/Warehouse.csv")
region = pd.read_csv("DATA/RAW/Region.csv")
```

```python
print("Customer Columns:")
print(customer.columns)

print("\nEmployee Columns:")
print(employee.columns)

print("\nOrders Columns:")
print(orders.columns)

print("\nOrderDetails Columns:")
print(orderdetails.columns)

print("\nProduct Columns:")
print(product.columns)

print("\nWarehouse Columns:")
print(warehouse.columns)

print("\nRegion Columns:")
print(region.columns)
```

# Dataset Description

## Customer.csv
Purpose:
Stores customer information such as Customer ID, Name, Address, Phone Number, etc.

---

## Employee.csv
Purpose:
Stores employee details who handle customer orders.

---

## Orders.csv
Purpose:
Stores order information such as Order ID, Customer ID, Order Date, Warehouse ID, etc.

---

## OrderDetails.csv
Purpose:
Stores product details for each order such as Product ID, Quantity and Price.

---

## Product.csv
Purpose:
Stores product information such as Product ID, Product Name, Category and Price.

---

## Warehouse.csv
Purpose:
Stores warehouse information where products are stored.

---

## Region.csv
Purpose:
Stores region information for warehouses.
